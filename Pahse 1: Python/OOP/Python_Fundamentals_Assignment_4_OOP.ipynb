## Q1. Classes & Objects — BankAccount

Create a class BankAccount with attributes: account_number, owner_name, and balance. Add methods to deposit, withdraw, and check balance.

**Your Solution:**

```python
class BankAccount:
    def __init__(self,account_number,owner_name,balance):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance
        
    #instance methods    
    def deposit(self,amount):
        self.balance += amount
        print(f"{amount} deposit succesfully! ")
        
    #instance method
    def withdraw(self,amount):
        self.balance -= amount
        print(f"{amount} withdraw successfully!")
    #instance method
    def check_balance(self):
        print(f"The balance is {self.balance}")
        
p1 = BankAccount(123,"Qouain",25_000)
p1.deposit(25_000)
p1.withdraw(13_000)
p1.check_balance()
```

## Q3. Encapsulation — Student

Create a class Student with attributes _name, _roll_no, and _marks. Provide private getter and setter methods with validation: marks cannot be negative, roll number must be between 1 and 100, and name cannot be empty.

**Your Solution:**

```python
class Student:
    def __init__(self,name,roll_no,marks):
        self.__name = name
        self.__roll_no = roll_no
        self.__marks = marks
        
    def getter(self):
        print(f"Name: {self.__name}, Roll No. {self.__roll_no} , marks: {self.__marks}")
        
    def setter(self,name,roll_no,marks):
        if name != "":
            self.__name = name
        else:
            print("Name cannot be empty")
    
        if roll_no >= 1 and roll_no <= 100:
            self.__roll_no = roll_no
        else:
            print("Roll number must be between 1 and 100")
                
        if marks >= 0:
            self.__marks = marks
        else:
            print("Marks cannot be negative")
               
s1 = Student("Qousain",23,50)
s1.getter()
s1.setter("",21,9)
s1.getter()
```

## Q5. Inheritance — Vehicle

Create a base class Vehicle with attributes brand and model. Create two subclasses, Car and Bike, that add extra attributes: seats (in Car) and engine_cc (in Bike).

**Your Solution:**

```python
class Vehicle:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
        
class Car(Vehicle):
    def __init__(self, brand, model,seats):
        super().__init__(brand, model)
        self.seats = seats
        
class Bike(Vehicle):
    def __init__(self,brand,model,engine_cc):
        super().__init__(brand,model)
        self.engine_cc = engine_cc
        
car1 = Car("Toyota", "Corolla", 5)
bike1 = Bike("Honda", "CB150F", 150)

print(car1.brand, car1.model, car1.seats)
print(bike1.brand, bike1.model, bike1.engine_cc)
```

## Q6. Abstraction — Employee

Create an abstract class Employee with an abstract method calculate_salary(). Create subclasses Intern, FullTimeEmployee, and ContractEmployee that implement the method differently.

**Your Solution:**

```python
from abc import ABC,abstractmethod

class Employee(ABC):
    
    @abstractmethod
    def calculate_salary(self):
        pass
    
class Inter(Employee):
    def calculate_salary(self):
        return 20000
        
class ContractEmployee(Employee):  
    def calculate_salary(self):
        return 50000
    
class FullTimeEmployee(Employee):
    def calculate_salary(self):
        return 100000
    
intern = Inter()
print("Intern Salary:", intern.calculate_salary())
contract = ContractEmployee()
full_time = FullTimeEmployee()
print("Contract Employee Salary:", contract.calculate_salary())
print("Full-Time Employee Salary:", full_time.calculate_salary())
    
```

## Q7. Constructor Overloading with Default Parameters — Person

Create a class Person whose constructor can work with: name only; name + age; name + age + address. Python does not support direct constructor overloading with multiple constructors, so use default parameters to simulate constructor overloading.

**Your Solution:**

```python
class Person:
   
    def __init__(self,name,age = None,address = None):
        self.name = name
        self.age = age
        self.address = address
        
    def display(self):
        print("Name:", self.name) 
        print("Age:", self.age) 
        print("Address:", self.address)
        

p1 = Person("Ali") 
p2 = Person("Ahmed", 20) 
p3 = Person("Usman", 21, "Peshawar")
p1.display() 
print() 
p2.display() 
print() 
p3.display()
```
