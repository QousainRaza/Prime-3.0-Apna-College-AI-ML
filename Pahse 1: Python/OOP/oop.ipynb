### Methods
- Instance methods : 1st Compulsory parameter self , they can access the class and instance attributes both

```python
class Laptop:
    storage_type = "SSD"
    def __init__(self,ram,storage):
        self.ram = ram
        self.storage = storage

    def get_info(self):
        print(f"Laptop has {self.ram}GB ram and {self.storage} {self.storage_type} storage.")
        
lap1 = Laptop(16,256)
lap1.get_info()
```

- Class methods: 1st parameter cls, access only class attributes , decorater we used '@classmethods'

```python
class Laptop:
    storage_type = "SSD"
    def __init__(self,ram,storage):
        self.ram = ram
        self.storage = storage
            
    @classmethod #classmethod decorator for classmethods
    def get_storage_type(cls): # class methods
        print(f"Storage type is {cls.storage_type}")
    
    def get_info(self):
        print(f"Laptop has {self.ram}GB ram and {self.storage} {self.storage_type} storage.")
        
lap1 = Laptop(16,256)
lap1.get_info()
Laptop.get_storage_type()
lap1.get_storage_type()
```

- Static methods: No compulsory parameters, instance and class attributes both were not accessed, @staticmethod decorator we used

```python
class Laptop:
    storage_type = "SSD"
    def __init__(self,ram,storage):
        self.ram = ram
        self.storage = storage
            
    @classmethod #classmethod decorator for classmethods
    def get_storage_type(cls): # class methods
        print(f"Storage type is {cls.storage_type}")
    
    def get_info(self):
        print(f"Laptop has {self.ram}GB ram and {self.storage} {self.storage_type} storage.")
        
    @staticmethod
    def calc_discount(price,discount):
        final_price = price - (discount* price/100)
        print(f"discounted price {final_price}")
        
lap1 = Laptop(16,256)
lap1.get_info()
Laptop.get_storage_type()
lap1.get_storage_type()
lap1.calc_discount(40_000,10)
```

Pracice Problem:
Product store

```python
class Product_store:
    count = 0
    def __init__(self,name,price):
        self.name = name
        self.price = price
        Product_store.count += 1
        
    def get_info(self):
        print(f"{self.name} price is Rs.{self.price}")
        
    @staticmethod
    def cal_discount(price,discount = 5):
        final_price = price - (discount* price/100)
        print(f"discounted price {final_price}")
        
    @classmethod
    def get_count(cls):
        print(f"Total count is {cls.count}")
        
p1 = Product_store("Laptop",95_000)
p1.get_info()
p1.cal_discount(p1.price)
Product_store.get_count()
```

#### 1. Encapsulation
- Wraping data & function into single units
- Data hiding
- Private attributes are accesed inside class
- Default public attributes created
- protected -> inside class + sub-classes 

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name = name
        self.__balance = balance
        
    def get_balance(self): #Getter
        return self.__balance
        
    def set_balance(self,newbalance): #Setter
        self.__balance = newbalance
            
acc1 = BankAccount("Qousain", 10)
# acc1.__balance
print(acc1.get_balance())
acc1.set_balance(50_000)
print(acc1.get_balance())
print(acc1._BankAccount__balance)
```

#### 2. Inheritance
- Resunig attr & methods from a Parent(Base) class.
- Child (Derive) class use the properties od base class

```python
class Employee:
    start_time = "10am"
    end_time = "6pm"

class Teacher(Employee):
     def __init__(self,subject):
         self.subject = subject
        
t1 = Teacher("Math")

print(t1.subject,t1.start_time,t1.end_time)
```

#### Types of Inheritence 
1. Single level: Employee -> AdminStaff
2. Multi level: Employee -> AdminStaff -> Accountant
3. Multiple Inheritence : ([Teacher] + [Student]) -> [TA]


```python
# Multi Level Inheritence
class Employee:
    start_time = "10am"
    end_time = "6pm"

class adminStaff(Employee):
    def __init__(self,role):
        self.role = role
    
class Accountant(adminStaff):
    def __init__(self, role,salary):
        super().__init__(role)
        self.salary = salary
        
acc1  = Accountant("CA",25_000)
print(acc1.start_time,acc1.end_time,acc1.salary,acc1.role)
```

```python
# Multiple Inheritence

class teacher:
    def __init__(self,salary):
        self.salary = salary
    
class student:
    def __init__(self,gpa):
        self.gpa = gpa
        
class TA(teacher,student):
    def __init__(self, salary,gpa,name):
        super().__init__(salary)
        student.__init__(gpa)
        self.name = name

ta1 = TA(25_000,3.0,"Qousain")
print(ta1.salary)
```

#### 3. Abstraction
- Hiding internal details & showing only essential features
- Abstract classes are part of abc module.
- blueprint for other classes

```python
from abc import ABC,abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class lion(Animal):
    def sound(self):
        print("Roar")
        
class cow(Animal):
    def sound(self):
        print("Mooo")
    
Lion = lion()
Lion.sound()
Cow = cow()
Cow.sound()
```

Types:

```python
# Function overriding
class employee:
    def get_designation(self):
        print("Designation = Employee")
        
class teacher(employee):
    def get_designation(self):
       print("Designation = Teacher") 
    
emp = teacher()
emp.get_designation()
```

```python
# Duck Typing - walks like a duck & quaks like a duck
class teacher(employee):
    def get_designation(self):
       print("Designation = Teacher")

class accountant:
    def get_designation(self):
        print("desihnation = accounting")
        
t1 = teacher()
t1.get_designation()
acc1 = accountant()
acc1.get_designation()
```
