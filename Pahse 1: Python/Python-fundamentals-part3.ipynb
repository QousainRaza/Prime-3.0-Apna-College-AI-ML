### String
- string are mutable

```python
word1 = "I love "
word2 = "Python"
# concatenation
print(word1 + word2)
#indexing
word2[2]

for ch in word1:
    print(ch)

# string are mutable i,e word1[2] = 'f' not posssible
```

```python
# slicing [start:end:jump]

word = "Python"
word[2:]
```

```python
# formatting
# 1. format()
a = 5
b = 10
sum = a + b 
print("Language is {}".format("Python"))
print("sum of {} and {} is {}".format(a,b,sum))
print("sum of {1} and {0} is {2}".format(a,b,sum))
```

```python
# f-string -> literal string interpolation
a = 5
b = 10
print(f"sum of {a} and {b} is {a+b}")
```

### List


```python
# mutable sequence of values
marks = [89,78,34,99,3,"abc",3.14]
len(marks)
marks[2]
```

```python
# methods
marks = [89,78,34,99,3,3.14]
marks.append(100)
marks
marks.insert(0,1000)
marks

marks.sort()
marks

marks.reverse()
marks
```
