# Numpy

```python
import numpy as np
import time
```

```python
# Python list vs Numpy Array
size = 1_000_000_000

#python list
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Numpy array time {end - start}")

#Numpy Array
np_arr = np.array([py_list])
start = time.time()
sq_array = np_arr ** 2
end = time.time()
print(f"Numpy array time {end - start}")
```
