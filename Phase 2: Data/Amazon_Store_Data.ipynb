# Assignment No. 1

- Clean and Structure our Data

### Problems
1. Data is inconsistent -> mixed data type
2. Missing value:
            
3. Duplicate data

```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename,"r") as f:
        data =json.load(f)
        return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
# Clean and Structure the data
# 1. Data inconsisty solving 
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        raw_rating = str(user["rating"]).strip().lower() # " 5 " => "5" by strip()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating
        
        ''' 2. Handle missing Value
        If we max number of features avalaible some data is missing then we represent is with
        'None', else we delete the complete row.
        '''
        raw_age = user.get("age") #.get -> if value exist it returns value else it return None ddo not error
        if (raw_age == None):
            user["age"] = None
            
        '''3. Deduplication -> Filter out unique values'''
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
cleaned_data = clean_data(data)

print(type(cleaned_data))
print(cleaned_data)
```

# Assignment 2


- Getting meaningful insights from data

1. What's avg rating?
2. What % of users gave poor rating(<3)?

```python
def get_insight(data):
    
    total_rating = 0 
    for user in data:
        total_rating += float(user["rating"])
        
    print(f"Avg of rating is {total_rating/len(data)}")
    
    #Percentage of users gave poor rating
    poor_rating = 0
    for user in data:
        if (float(user["rating"]) < 3):
            poor_rating += 1
            
    print(f"percentage of poor rating is {poor_rating/len(data)*100}%")
        
```

```python
get_insight(cleaned_data)
```

# Assignment 3

- Build a product recommendation feature (Rule based, not ML)

1. If useer rating >= 4, recommend same brand products.
2. If user rating < 4, recommand different brand products.

```python
def get_recommandations(data):
    
    recommandation = []
    
    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        if float(user["rating"]) >= 4:
            curr_recomm["brand"] = "apple"
        else:
            curr_recomm["brand"] = "Samsung"
            
        recommandation.append(curr_recomm)
        
    return recommandation
            
```

```python
get_recommandations(cleaned_data)
```
