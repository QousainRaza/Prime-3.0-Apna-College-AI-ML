## Grouping and Aggregate of Data
- groupby()
- mean/sum/max/min/count
- agg()

```python
df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].mean()

df.groupby("country")["income"].agg(avg_salary = "mean",min_salary = "min",max_salary= "max",count= "count")
```

## Melt & Pivot

- Melt - wide to long format <br>
melt(id_vars #want to keep, value_vars #want unpivot, var_name #new_col name, value_name #new_col value)

- Pivot - long to wide format
pivot(index #keep, coulmns,values)

```python
df = pd.DataFrame({
    "country" : ["USA","USA","India","Pakistan"],
    "year" : [2020,2021,2020,2021],
    "sales" : [100,120,90,130],
    "profit" : [20,30,18,22],
    "tax" : [10,20,30,40]
})
```

```python
melted_df = df.melt(
    id_vars = ["country","year"],
    value_vars = ["sales","profit","tax"],
    var_name = "matrics",
    value_name = "value"
)
```

```python
melted_df
```

```python
original = melted_df.pivot(
    index = ["country","year"],
    columns= "matrics",
    values="value"
)
```

```python
original
```

## Merge and Join
- inner join
- left join 
- right join
- outer join

```python
df_customer = pd.DataFrame(
    {
        "customer_id" : [1,2,3,4],
        "name" : ["Adam","Bob","Charlie","Dave"]
    }
)

df_orders = pd.DataFrame({
    "order_id" : [101,102,103,104],
    "customer_id" : [2,1,4,5],
    "amount" : [250, 120, 300, 180]
})
```

```python
df_customer
```

```python
df_orders
```

```python
#inner join
pd.merge(df_customer,df_orders, on = "customer_id")
```

```python
#left join
pd.merge(df_customer,df_orders, on = "customer_id",how = "left")
```

```python
#right join
pd.merge(df_customer,df_orders, on = "customer_id",how = "right")
```

```python
#outer join
pd.merge(df_customer,df_orders, on = "customer_id",how = "outer")
```

## Data concatenation

```python
df1 = pd.DataFrame(
    {
        "id" : [1,2,3],
        "name" : ["Adam","Bob","Dave"]
    }
)

df2 = pd.DataFrame(
    {
        "id" : [4,5,6],
        "name" : ["Eve","Fester","Charlie"]
    }
)
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index=True)
```
