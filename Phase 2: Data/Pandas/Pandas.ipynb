### Filtering data using query methods - query() -> return copy not view
- Express as a string
- Column Based references
- Backtricks for col_name with spaces or spacial chars
- Operators (&,|,~,<,<=,>=,>,==,etc)
- @ to reference Python vars
- Chained Comparison



```python

aqi_val = 100
aqi_data = df.query("aqi > @aqi_val & temperature > 30")[["city","aqi"]].copy() # chaining [["city","aqi"]]
```
