# Data Cleaning

## 1. Handle Missing Values
- isnull()/isna()
- isnull().sum()
- dropna()
- fillna(value)
- ffill()
- bfill()

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df
```

```python
df.info()
```

```python
df.isnull()
```

```python
df.isna()
```

```python
df.isnull().sum()
```

```python
# drop row having missing value or empty cell
df.dropna()
```

```python
# drop columns having empty cell
df.dropna(axis = 1)
```

```python
# fill mising value with any value
df.fillna(100)
cleaned_data = df.copy()
cleaned_data["age"] = df["age"].fillna(df["age"].mean())
```

```python
cleaned_data
```

```python
df
```

```python
# forward fill
df.ffill()
```

```python
# backward fill
df.bfill()
```
