## Q1. Salary Tax Rate

Write a program that takes **salary** as input. Using conditional statements, calculate the final tax rate based on these rules:

- If salary < 30,000 → 5%
- If salary is 30,000–70,000 → 15%
- If salary > 70,000 → 25%


```python
salary = float(input("salary: "))

if salary < 30000 :
    tex = salary * 5 / 100
    print("5% tex : ",tex)
elif salary >= 30000 and salary <= 70000 :
    tex = salary * 15 / 100
    print("15% tex : ",tex)
elif salary > 70000:
    tex = salary * 25 / 100
    print("25% tex : ",tex)
else:
    print("Invalid amount")
```

For example, for `9`, we check the range `(2, 8)` and it gets divided by `3`, so `9` is non-prime and we return `False`. For `7`, we check the range `(2, 6)` and it does not get divided by any number, so `7` is prime and we return `True`.
