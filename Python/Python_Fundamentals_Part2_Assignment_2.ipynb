## Q1. Salary Tax Rate

Write a program that takes **salary** as input. Using conditional statements, calculate the final tax rate based on these rules:

- If salary < 30,000 → 5%
- If salary is 30,000–70,000 → 15%
- If salary > 70,000 → 25%


```python
salary = float(input("salary: "))

if salary < 30000 :
    tex = salary * 5 / 100
    print("5% tex : ",tex)
elif salary >= 30000 and salary <= 70000 :
    tex = salary * 15 / 100
    print("15% tex : ",tex)
elif salary > 70000:
    tex = salary * 25 / 100
    print("25% tex : ",tex)
else:
    print("Invalid amount")
```

## Q3. Print Digits

Write a function that prints the **digits** of a number `n`.

For example, if `n = 312`, there are 3 digits in it: `3`, `1`, and `2`, and we need to print them.

**Hint:**
- The rightmost digit of a number `N` is `N % 10`.
- To remove the rightmost digit from a number, we can use `N = N / 10`.


```python
def digit(n):
    digits = []
    while n > 0:
        N = n % 10
        digits.append(N)
        n = n//10
    print(digits)

digit(312)
    
```

## Q4. Count Digits

Write a function to return the **number of digits** in a number `n`.


```python
def no_of_digits(n):
    count = 0
    while n > 0:
        n = n//10
        count+=1
    print(count)
no_of_digits(123344)
    
```

## Q5. Sum of Digits

Write a function to return the **sum of digits** of a number `n`.


```python
def sum_digits(n):
    sum = 0
    while n > 0:
        N = n % 10
        sum += N
        n = n // 10
    print(sum)
    
sum_digits(677)
        
```

## Q7. Positive or Negative

Design a program to continuously input a number from the user and print if it is **positive or negative** until the user enters `"Quit"`.


```python
def check_number():
    while True:
        num = input("number or Quit: ")
        if num == 'Quit':
            break
        num = int(num)
        if num >= 0:
            print("Positive Number!")
        else:
            print("Negative Number!")
            
check_number()
```

## Q9. Prime Number

Write a function `is_prime(n)` that returns **True** if `n` is a prime number and **False** otherwise, using a loop.

**Hints:**
1. We only check prime for 2 or numbers greater than 2. `2` is the smallest prime number.
2. A non-prime number `n` will always get divided by at least one number in the range `[2, n-1]`.

For example, for `9`, we check the range `(2, 8)` and it gets divided by `3`, so `9` is non-prime and we return `False`. For `7`, we check the range `(2, 6)` and it does not get divided by any number, so `7` is prime and we return `True`.


```python
def is_prime(n):
    for i in range(2,n-1):
        if n % i == 0:
            return False
    return True
is_prime(7)
```

## Q10. Number Guessing Game

Let’s create a **Number Guessing Game**. Given a secret number (already decided by you), write a program that asks the user to guess it and prints:

- If the guess is above the number → **"Too high"**
- If the guess is below the number → **"Too low"**
- If the guess matches → print an appropriate success message.


```python
def number_guessing_game():
    n = 10
    while True:
        num = int(input("Enter number to guess: "))
        if num > n: 
            print("Too high")
        elif num < n :
            print("Too low")
        else:
            print("match, WIN")
            break
        
number_guessing_game() 
        
```
