## Conditional statements


```python
color = input("color: ")

if color == 'red':
    print("stop")
elif color == 'green':
    print("go")
else:
    print( "wrong"
    )
```

```python
age = int(input("Age: "))
if age < 13 :
    print("child")
elif age > 13 and age <= 18:
    print("teenage")
else:
    print("adult")
```

```python
num = int(input("num: "))

if num%2==0 :
    print("even")
else:
    print("odd")
```

## Loops


### While loop

```python
count = 0
while count<5:
    print("hello world")
    count+=1
```

```python
i = 5
while i>=1:
    print(i)
    i-=1
    
```

```python
n = int(input("Enter n: "))

count = 1
while(count <= 10):
    result = n * count
    print(n, "x", count ,"=" , result )
    count+=1
```

### break and continue

```python
# break -> terminate loop
```

```python
i = 1

while i <10:
    if(i%6==0):
        break
    print(i)
    i+=1
    
print("outside loop")
```

```python
# continue -> skip current iteration and. ove to next iteration
```

```python
# odd numbers using continue
i = 0
while i<10:
    i+=1
    if(i%2 == 0):
        continue
    print(i)
print("outside loop")
```

### for loops

```python
# for loop -> sequentila loop

string = "hello"
for var in string:
    print(var)

# in -> membership operator -> check presence

string = "hello"
if 'o' in string:
    print("o is exist string")
```

```python
word = "Artificial Intelligence"
count = 0

for ch in word:
    if ch == 'i' or ch == 'I' :
        count+=1
print("count of i in word is ",count )
```

```python
# range(5) - 0 to 4
# range(start,stop,step) -> stop is mendatory
```

```python
n = int(input("n: "))
sum = 0
for i in range(1, n+1):
    sum+=i
print(sum)
```

## Function

```python
def sum(a,b):   #a,b parameters
    sum = a+b
    return sum

sum(4,4)    #4,4 arguments
```

```python
def avg(a,b,c):
    return (a+b+c)/3

avg(3,3,3)
```

```python
# function(non-default parameters,default parameters)
def sum(a,b=1):
    return a+b
sum(1)
```

```python
def fact(n):
    fact = 1
    if (n == 1 or n == 0):
        return fact
    else:
        for i in range(1,n+1):
            fact =fact * i
        return fact
        
n = int(input("n : "))
print(fact(n))
```
