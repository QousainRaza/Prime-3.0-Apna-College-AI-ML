### Conditional statements


```python
color = input("color: ")

if color == 'red':
    print("stop")
elif color == 'green':
    print("go")
else:
    print( "wrong"
    )
```

```python
age = int(input("Age: "))
if age < 13 :
    print("child")
elif age > 13 and age <= 18:
    print("teenage")
else:
    print("adult")
```

```python
num = int(input("num: "))

if num%2==0 :
    print("even")
else:
    print("odd")
```
